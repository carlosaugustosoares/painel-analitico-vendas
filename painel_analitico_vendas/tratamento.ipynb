```python
# Tratamento da base de vendas
# Python + Pandas
```

```python
# Importando o pandas
import pandas as pd
```

```python
# Carregando a base
df = pd.read_csv('base_vendas.csv', encoding='utf-8-sig')

print('Base carregada!')
display(df.head())
```

```python
# Entendendo a base
print('\n--- DIMENSÕES DA BASE ---')
print('Linhas:', df.shape[0])
print('Colunas:', df.shape[1])

print('\n--- COLUNAS ---')

display(df.columns)

print('\n--- TIPOS DE DADOS ---')

display(df.dtypes)
```

```python
# Verificando valores ausentes
print('\n ---VALORES AUSENTES ---')
display(df.isnull().sum())
```

```python
# Corrigindo tipos de dados

# Data
df['Data_Venda'] = pd.to_datetime(
    df['Data_Venda'],
    errors='coerce'
)

# Números
df['Quantidade'] = pd.to_numeric(
    df['Quantidade'],
    errors='coerce'
)

df['Preco_Unitario'] = pd.to_numeric(
    df['Preco_Unitario'],
    errors='coerce'
)

df["Faturamento"] = pd.to_numeric(
    df["Faturamento"],
    errors='coerce'
)
```

```python
# Padronizando textos
df['Produto'] = df['Produto'].str.strip()
df['Categoria'] = df['Categoria'].str.strip()
df['Regiao'] = df['Regiao'].str.strip()
df['Forma_Pagamento'] = (df['Forma_Pagamento'].str.strip())
df['Cliente'] = df['Cliente'].str.strip()
```

```python
# Tratando valores ausentes

# Quantidade
# Usaremos a mediana para preencher os valores ausentes.

mediana_quantidade = df['Quantidade'].median()

df['Quantidade'] = df['Quantidade'].fillna(
    mediana_quantidade
)

# Região
df['Regiao'] = df['Regiao'].fillna(
    'Não informado'
)

# Forma de pagamento
df['Forma_Pagamento'] = (
    df['Forma_Pagamento'].fillna('Não informado')
)
```

```python
# Tratando valores ausentes

print('\nDatas inválidas ou ausentes:')

print(
    df['Data_Venda'].isnull().sum()
)

# Como não é possível saber a data correta, vamos remover essas linhas.

df = df.dropna(
    subset=['Data_Venda']
)
```

```python
# Verificando valores duplicados

print('\nDuplicados encontrados:')

print(
    df.duplicated().sum()
)

# Removendo os duplicados

df = df.drop_duplicates()
```

```python
# Recalculando o faturamento

df['Faturamento'] = (
    df['Quantidade'] *
    df['Preco_Unitario']
)
```

```python
# Criando colunas auxiliares

df['Ano'] = df['Data_Venda'].dt.year

df['Mes'] = df['Data_Venda'].dt.month

df['Mes_Ano'] = (
    df['Data_Venda'].dt.to_period('M').astype(str)
)
```

```python
# Validar a base novamente

print('\n--- BASE DEPOIS DO TRATAMENTO ---')

display(df.head())


print('\n--- VALORES AUSENTES ---')

display(df.isnull().sum())


print('\n--- TIPOS DE DADOS ---')

display(df.dtypes)


print('\n--- DIMENSÕES FINAIS ---')

print('Linhas:', df.shape[0])
print('Colunas:', df.shape[1])
```

```python
# Salvando a base limpa

df.to_csv(
    'vendas_limpa.csv',
    index=False,
    encoding='utf-8-sig'
)

print('\nBase limpa salva como:')
print('vendas_limpa.csv')
```
